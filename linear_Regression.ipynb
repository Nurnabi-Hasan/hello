```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv('Salary.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
plt.scatter(df['YearsExperience'], df ['Salary'], color='green')
plt.xlabel('Year Of Experiance')
plt.ylabel('Salary')
plt.title('Salary Prediction')
```

```python
x = df.drop(columns=['Salary'])
x.head()
```

```python
y = df.drop(columns=['YearsExperience'])

y.head()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(x,y) # train ml model with x and y data
```

```python
reg.predict([[20]]) # final prediction result
```

```python
plt.scatter(x.mean(), y.mean(), color='red')

plt.xlabel('Year Of Experiance')
plt.ylabel('Salary')
plt.title('Salary Prediction')
```

```python
plt.plot(x, reg.predict(x))
plt.scatter(x.mean(), y.mean(), color='red')
plt.scatter(df['YearsExperience'], df ['Salary'], color='green')
plt.xlabel('Year Of Experiance')
plt.ylabel('Salary')
plt.title('Salary Prediction')
```

```python
reg.score(x,y) #check the accurcu of the model
```

loss and Cost Function

```python
df['predict_s'] = reg.predict(x)
df.head()
```

```python
df['lose'] = df['Salary'] - df['predict_s']
df.head()
```

```python
#for mSe
from sklearn.metrics import mean_squared_error
```
